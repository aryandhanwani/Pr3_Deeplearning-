### Task 1: Environment Setup, Image I/O & Morphological Operations

1.1 Environment setup & image loading

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
print("OpenCV version:", cv2.__version__)
```

```python
def show(img, title):
    plt.figure(figsize=(5,5))

    if len(img.shape) == 3:
        img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

    plt.imshow(img, cmap='gray')
    plt.title(title)
    plt.axis('off')
    plt.show()
```

```python
img2 = cv2.imread('D:\Deep learning\Pr.3\Assets\Color_Owl.jpg')

show(img2, "BGR2RGB Image")
```

1.2 Binarize a test image

```python
to_grayscale = cv2.cvtColor(img2, cv2.COLOR_BGR2GRAY)

_, to_threshold = cv2.threshold(to_grayscale, 127, 255, cv2.THRESH_BINARY)

plt.figure(figsize=(10,5))

plt.subplot(1, 2, 1)
plt.imshow(to_grayscale, cmap='gray')
plt.title("Grayscale Image")
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(to_threshold, cmap='gray')
plt.title("Threshold Image")
plt.axis('off')

plt.show()
```

1.3 Apply the four morphological operations

```python
kernel = np.ones((5,5), np.uint8)
```

```python
erosion = cv2.erode(to_threshold, kernel, iterations=1)
dilation = cv2.dilate(to_threshold, kernel, iterations=1)
opening = cv2.morphologyEx(to_threshold, cv2.MORPH_OPEN, kernel)
closing = cv2.morphologyEx(to_threshold, cv2.MORPH_CLOSE, kernel)
```

```python
plt.figure(figsize=(15,10))

plt.subplot(2, 2, 1)
plt.imshow(erosion, cmap='gray')
plt.title("Erosion")
plt.axis('off')

plt.subplot(2, 2, 2)
plt.imshow(dilation, cmap='gray')
plt.title("Dilation")
plt.axis('off')

plt.subplot(2, 2, 3)
plt.imshow(opening, cmap='gray')
plt.title("Opening")
plt.axis('off')

plt.subplot(2, 2, 4)
plt.imshow(closing, cmap='gray')
plt.title("Closing")
plt.axis('off')

plt.tight_layout()
plt.show()
```

1.4 Kernel shape & size comparison

```python
kernel_rect_3 = cv2.getStructuringElement(cv2.MORPH_RECT, (3,3))
kernel_rect_5 = cv2.getStructuringElement(cv2.MORPH_RECT, (5,5))
kernel_rect_9 = cv2.getStructuringElement(cv2.MORPH_RECT, (9,9))

kernel_elipse_3 = cv2.getStructuringElement(cv2.MORPH_ELLIPSE, (3,3))
kernel_elipse_5 = cv2.getStructuringElement(cv2.MORPH_ELLIPSE, (5,5))
kernel_elipse_9 = cv2.getStructuringElement(cv2.MORPH_ELLIPSE, (9,9))

kernel_cross_3 = cv2.getStructuringElement(cv2.MORPH_CROSS, (3,3))
kernel_cross_5 = cv2.getStructuringElement(cv2.MORPH_CROSS, (5,5))
kernel_cross_9 = cv2.getStructuringElement(cv2.MORPH_CROSS, (9,9))
```

```python
banana_noisy = cv2.imread('D:\Deep learning\Pr.3\Assets\Banana_noisy_binary.jpg')
to_grayscale_banana = cv2.cvtColor(banana_noisy, cv2.COLOR_BGR2GRAY)
_, to_threshold_banana = cv2.threshold(to_grayscale_banana, 127, 255, cv2.THRESH_BINARY)
```

```python
opening_rect_3 = cv2.morphologyEx(to_threshold_banana, cv2.MORPH_OPEN, kernel_rect_3)
opening_rect_5 = cv2.morphologyEx(to_threshold_banana, cv2.MORPH_OPEN, kernel_rect_5)
opening_rect_9 = cv2.morphologyEx(to_threshold_banana, cv2.MORPH_OPEN, kernel_rect_9)

closing_rect_3 = cv2.morphologyEx(to_threshold_banana, cv2.MORPH_CLOSE, kernel_rect_3)
closing_rect_5 = cv2.morphologyEx(to_threshold_banana, cv2.MORPH_CLOSE, kernel_rect_5)
closing_rect_9 = cv2.morphologyEx(to_threshold_banana, cv2.MORPH_CLOSE, kernel_rect_9)

opening_elipse_3 = cv2.morphologyEx(to_threshold_banana, cv2.MORPH_OPEN, kernel_elipse_3)
opening_elipse_5 = cv2.morphologyEx(to_threshold_banana, cv2.MORPH_OPEN, kernel_elipse_5)
opening_elipse_9 = cv2.morphologyEx(to_threshold_banana, cv2.MORPH_OPEN, kernel_elipse_9)

closing_elipse_3 = cv2.morphologyEx(to_threshold_banana, cv2.MORPH_CLOSE, kernel_elipse_3)
closing_elipse_5 = cv2.morphologyEx(to_threshold_banana, cv2.MORPH_CLOSE, kernel_elipse_5)
closing_elipse_9 = cv2.morphologyEx(to_threshold_banana, cv2.MORPH_CLOSE, kernel_elipse_9)

opening_cross_3 = cv2.morphologyEx(to_threshold_banana, cv2.MORPH_OPEN, kernel_cross_3)
opening_cross_5 = cv2.morphologyEx(to_threshold_banana, cv2.MORPH_OPEN, kernel_cross_5)
opening_cross_9 = cv2.morphologyEx(to_threshold_banana, cv2.MORPH_OPEN, kernel_cross_9)

closing_cross_3 = cv2.morphologyEx(to_threshold_banana, cv2.MORPH_CLOSE, kernel_cross_3)
closing_cross_5 = cv2.morphologyEx(to_threshold_banana, cv2.MORPH_CLOSE, kernel_cross_5)
closing_cross_9 = cv2.morphologyEx(to_threshold_banana, cv2.MORPH_CLOSE, kernel_cross_9)
```

### Task 2: Bitwise Operations & Image Histograms

2.1 Bitwise AND / OR / XOR / NOT

```python
mask_circle = np.zeros((300, 300), dtype=np.uint8)
mask_rectangle = np.zeros((300, 300), dtype=np.uint8)

cv2.circle(mask_circle, (150, 150), 80, 255, -1)
cv2.rectangle(mask_rectangle, (75, 75), (225, 225), 255, -1)
```

```python
bitwise_and = cv2.bitwise_and(mask_circle, mask_rectangle)
bitwise_or = cv2.bitwise_or(mask_circle, mask_rectangle)
bitwise_xor = cv2.bitwise_xor(mask_circle, mask_rectangle)
bitwise_not = cv2.bitwise_not(mask_circle,)
```

```python
plt.figure(figsize=(15,10))

plt.subplot(2, 3, 1)
plt.imshow(mask_circle, cmap='gray')
plt.title("Circle Mask")
plt.axis('off')

plt.subplot(2, 3, 2)
plt.imshow(mask_rectangle, cmap='gray')
plt.title("Rectangle Mask")
plt.axis('off')

plt.subplot(2, 3, 3)
plt.imshow(bitwise_and, cmap='gray')
plt.title("Bitwise AND")
plt.axis('off')

plt.subplot(2, 3, 4)
plt.imshow(bitwise_or, cmap='gray')
plt.title("Bitwise OR")
plt.axis('off')

plt.subplot(2, 3, 5)
plt.imshow(bitwise_xor, cmap='gray')
plt.title("Bitwise XOR")
plt.axis('off')

plt.subplot(2, 3, 6)
plt.imshow(bitwise_not, cmap='gray')
plt.title("Bitwise NOT")
plt.axis('off')

plt.tight_layout()
plt.show()
```

### Task 3: Face Detection with YuNet (OpenCV 5.0)


3.1 Load the YuNet face detector

```python
person = cv2.imread(r'D:\Deep learning\Pr.3\Assets\Person.png')

img_h, img_w = person.shape[:2]
```

```python
model_path = r'D:\Deep learning\Pr.3\Assets\face_detection_yunet_2023mar.onnx'

detector = cv2.FaceDetectorYN_create(model_path, '', (img_w, img_h), score_threshold=0.9)
```

```python
detect = detector.setInputSize((img_w, img_h)) 
```

```python
_, faces = detector.detect(person)
print(faces)
```

```python
person2 = cv2.imread(r'D:\Deep learning\Pr.3\Assets\Person_2.png')
person3 = cv2.imread(r'D:\Deep learning\Pr.3\Assets\Person_3.jpg')
```

```python
img_h, img_w = person.shape[:2]
detector.setInputSize((img_w, img_h))
_, faces1 = detector.detect(person)

img_h, img_w = person2.shape[:2]
detector.setInputSize((img_w, img_h))
_, faces2 = detector.detect(person2)

img_h, img_w = person3.shape[:2]
detector.setInputSize((img_w, img_h))
_, faces3 = detector.detect(person3)
```

```python
for face in faces1:

    x, y, w, h = face[:4].astype(int)

    cv2.rectangle(person, (x, y), (x+w, y+h), (0, 255, 0), 1)

    landmarks = face[4:14].reshape(5, 2).astype(int)

    for point in landmarks:
        cv2.circle(person, tuple(point), 3, (0, 0, 255), -1)

    confidence = face[14]
    cv2.putText(person, str(round(confidence, 2)), (x, y-10),cv2.FONT_HERSHEY_SIMPLEX, 0.6, (0, 255, 0), 2)

show(person, "Face Detection - Person 1")
```
